```python
import base64
from dotenv import load_dotenv
from openai import OpenAI
import base64

load_dotenv()

client = OpenAI()
```

```python
MODEL="gpt-4o"
```

```python

def encode_image(image_path):    
    with open(image_path, "rb") as image_file:        
        return base64.b64encode(image_file.read()).decode("utf-8")
```

```python
def get_code_page(path_new_image) -> str:

    image1 = encode_image("/home/matheus/Documentos/mestrado-matheus-lisboa/dataset-generation/imgs/exemplo1.png")
    image2 = encode_image("/home/matheus/Documentos/mestrado-matheus-lisboa/dataset-generation/imgs/exemplo2.png")

    new_image = encode_image(path_new_image)


    with open("/home/matheus/Documentos/mestrado-matheus-lisboa/experimento2/handmade_html_dataset/english/adição/example11.html", "r") as file:
        example1 = file.read()
    
    with open("/home/matheus/Documentos/mestrado-matheus-lisboa/experimento2/handmade_html_dataset/english/adição/example14.html", "r") as file:
        example2 = file.read()

    PROMPT1 = f"""
Sua tarefa vai ser gerar um código HTML que descreva a imagem 3, a partir dos exemplos das imagens 1 e 2.

Inclua também o cabeçalho do exemplo, bem como o JSON explicando a atividade
Utilize a mesma forma de descrever que as dos exemplos:


Exemplo de descrição da imagem 1:

{example1}

Exemplo de descrição da imagem 2:

{example2}

Gere a atividade que descreva a imagem 3.

"""

    response = client.chat.completions.create(
        model=MODEL,    
        messages=[
            {"role": "system", "content": "você é um assistente que ajuda a converter uma imagem em código."},
            {"role": "user", "content": [
                {"type": "text", "text": PROMPT1},
                {"type": "image_url", "image_url": {
                        "url": f"data:image/png;base64,{image1}"}            
                },
                {"type": "image_url", "image_url": {
                        "url": f"data:image/png;base64,{image2}"}            
                },
                {"type": "image_url", "image_url": {
                        "url": f"data:image/png;base64,{new_image}"}            
                }       
            ]}    
            ],
        temperature=0.0
    )

    return response.choices[0].message.content
```

```python
text = get_code_page("/home/matheus/Documentos/mestrado-matheus-lisboa/dataset-generation/imgs/nova.png")
with open("examples/example1.html", "w") as file:
    file.write(text)
```
